# **Anàlisi de les Pluges per Estació**

Aquest notebook analitza l'evolució mensual de les pluges a tres estacions meteorològiques:
- **Sabadell Centre**
- **Sabadell Nord (Parc Agrari)**
- **Vacarisses**

Per a cada estació es mostra:
1. L'evolució anual de la pluja per a cada mes (gràfic 4×3)
2. La mitjana mensual de pluja al llarg de tot el període
3. Una comparació final entre les tres estacions

## 1. Configuració i Funcions Auxiliars

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import warnings

warnings.filterwarnings('ignore')

month_names = {
    1: 'Gener', 2: 'Febrer', 3: 'Març', 4: 'Abril',
    5: 'Maig', 6: 'Juny', 7: 'Juliol', 8: 'Agost',
    9: 'Setembre', 10: 'Octubre', 11: 'Novembre', 12: 'Desembre'
}
```

```python
def load_station_data(url):
    """Carrega les dades diàries d'una estació i afegeix columnes de temps."""
    df = pd.read_csv(url)
    if 'Unnamed: 0' in df.columns:
        df.drop(columns=['Unnamed: 0'], inplace=True)
    df['year'] = pd.to_datetime(df['date']).dt.year
    df['month'] = pd.to_datetime(df['date']).dt.month
    df['day'] = pd.to_datetime(df['date']).dt.day
    return df


def get_monthly_series(df):
    """Retorna un diccionari {mes: sèrie de pluja anual} per als 12 mesos."""
    monthly_totals = df.groupby(['year', 'month'])['rain_mm'].sum()
    return {m: monthly_totals.xs(m, level='month') for m in range(1, 13)}


def plot_monthly_grid(monthly_series, station_name):
    """Gràfic 4×3 amb l'evolució anual de la pluja per a cada mes."""
    fig, axes = plt.subplots(4, 3, figsize=(15, 10), sharex=False)
    axes = axes.flatten()

    for i in range(12):
        ax = axes[i]
        data = monthly_series[i + 1]
        data.plot.bar(ax=ax, color='#2980b9')
        ax.set_title(month_names[i + 1], fontsize=12, fontweight='bold')
        ax.set_ylim(0, 200)
        ax.xaxis.set_major_locator(MaxNLocator(integer=True))

        mean_val = data.mean()
        ax.axhline(mean_val, color='red', linestyle='--', linewidth=1)
        ax.text(
            ax.get_xlim()[1],
            mean_val,
            f'{mean_val:.1f}',
            va='center',
            ha='left',
            fontsize=8,
            color='red',
        )

    fig.suptitle(f'Evolució Anual de Pluja — {station_name}', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()


def plot_monthly_average(df, station_name, period):
    """Gràfic de barres amb la mitjana mensual de pluja."""
    monthly_totals = df.groupby(['year', 'month'])['rain_mm'].sum()
    month_rain = monthly_totals.groupby('month').mean()
    month_rain.index = month_rain.index.map(month_names)

    fig, ax = plt.subplots(figsize=(10, 5))
    month_rain.plot.bar(ax=ax, color='#2980b9')
    # Afegir valors a les barres
    for bar in ax.patches:
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}', ha='center', va='bottom', fontsize=9)
    ax.set_xlabel('Mes', fontsize=12)
    ax.set_ylabel('Pluja (mm)', fontsize=12)
    ax.set_title(f'Mitjana Mensual de Pluja — {station_name} ({period})', fontsize=14, fontweight='bold')
    ax.tick_params(axis='x', rotation=45)
    plt.tight_layout()
    plt.show()

    return month_rain
```

## 7. Mètodes Alternatius de Detecció de Tendències

A continuació es comparen els resultats del Seasonal Mann-Kendall amb tres enfocaments alternatius:

1. **STL + Mann-Kendall** — Descomposició STL (Seasonal-Trend decomposition using LOESS) del total mensual, i Mann-Kendall sobre el component de tendència.
2. **GAM (Splines) + bootstrap** — Regressió amb splines cúbiques naturals per modelar la tendència suau amb IC del 95% per bootstrap.
3. **SARIMA** — Model ARIMA estacional amb tendència lineal, per avaluar si l'autocorrelació afecta la significança.

L'objectiu és validar si la tendència detectada pel Seasonal Mann-Kendall a Sabadell és robusta a la metodologia.

```python
# === 7.1 STL Decomposition + Mann-Kendall ===
from statsmodels.tsa.seasonal import STL
import matplotlib.pyplot as plt

# Prepare monthly time series (continuous index)
def build_monthly_series(df, name):
    monthly = df.groupby(['year', 'month'])['rain_mm'].sum().reset_index()
    monthly['date'] = pd.to_datetime(monthly['year'].astype(str) + '-' + monthly['month'].astype(str) + '-01')
    monthly = monthly.sort_values('date').set_index('date')
    return monthly['rain_mm']

monthly_series = {}
for name, df in [('Sabadell Centre', df_sbd_centre), ('Sabadell Nord', df_sbd_nord), ('Vacarisses', df_vac)]:
    monthly_series[name] = build_monthly_series(df, name)
    print(f"{name}: {len(monthly_series[name])} mesos")
```

```python
# STL decomposition + MK on trend for each station
stl_results = {}
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

for idx, (name, series) in enumerate(monthly_series.items()):
    # STL decomposition
    stl = STL(series, period=12, robust=True)
    res = stl.fit()
    stl_results[name] = res

    # Mann-Kendall on trend component
    trend_vals = res.trend.dropna().values
    years_idx = np.arange(len(trend_vals))
    mk = mann_kendall(trend_vals)
    ss = sens_slope(trend_vals)

    # Plot
    ax = axes[idx]
    ax.plot(series.index, series.values, 'o-', color='lightblue', markersize=3, label='Original', alpha=0.5)
    ax.plot(res.trend.index, res.trend.values, 'r-', linewidth=2.5, label='STL Trend')
    ax.fill_between(res.trend.index,
                     res.trend.values - res.resid.std(),
                     res.trend.values + res.resid.std(),
                     alpha=0.2, color='red', label='±1 residu std')
    ax.set_title(f'{name} — STL Trend: tau={mk["tau"]:.3f}, p={mk["p_value"]:.4f} ({mk["trend"]})', fontsize=12)
    ax.set_ylabel('Pluja (mm)')
    ax.legend(fontsize=8)
    ax.grid(alpha=0.3)

plt.xlabel('Data')
plt.tight_layout()
plt.show()

# Print summary table
print(f"\n{'='*80}")
print(f"  STL + Mann-Kendall — Comparison with Seasonal MK")
print(f"{'='*80}")
print(f"{'Station':<20} {'MK tau':>8} {'p-value':>10} {'MK trend':<14} {'Sen slope':>10} {'SMK p-value':>14} {'SMK trend':<14}")
print("-" * 90)
smk_pvalues = {'Sabadell Centre': 0.0024, 'Sabadell Nord': 0.0138, 'Vacarisses': 0.5966}
smk_trends = {'Sabadell Centre': 'increasing', 'Sabadell Nord': 'increasing', 'Vacarisses': 'no trend'}
for name in monthly_series:
    res = stl_results[name]
    mk = mann_kendall(res.trend.dropna().values)
    ss = sens_slope(res.trend.dropna().values)
    print(f"{name:<20} {mk['tau']:>8.4f} {mk['p_value']:>10.4f} {mk['trend']:<14} {ss['slope']:>10.2f} {smk_pvalues[name]:>14.4f} {smk_trends[name]:<14}")
```

### Interpretació STL + MK

L'STL descompon la sèrie mensual en **tendència + estacionalitat + residu**. Aplicant Mann-Kendall només al component de tendència (lliure d'estacionalitat) obtenim una validació independent del Seasonal Mann-Kendall.

Si ambdós mètodes coincideixen, la confiança en la tendència detectada augmenta.

```python
# === 7.2 GAM-like Trend with Natural Splines + Bootstrap CI ===
from scipy.interpolate import UnivariateSpline

def trend_with_bootstrap(series, n_bootstrap=1000, ci=0.95):
    """Fit spline trend + bootstrap confidence intervals."""
    x = np.arange(len(series))
    y = series.values
    mask = ~np.isnan(y)
    x, y = x[mask], y[mask]

    s = UnivariateSpline(x, y, s=len(x) * 2)
    y_fit = s(x)

    # Bootstrap residuals
    residuals = y - y_fit
    boot_fits = np.zeros((n_bootstrap, len(x)))
    for b in range(n_bootstrap):
        boot_res = np.random.choice(residuals, size=len(x), replace=True)
        boot_y = y_fit + boot_res
        boot_s = UnivariateSpline(x, boot_y, s=len(x) * 2)
        boot_fits[b] = boot_s(x)

    alpha = (1 - ci) / 2
    ci_lower = np.percentile(boot_fits, alpha * 100, axis=0)
    ci_upper = np.percentile(boot_fits, (1 - alpha) * 100, axis=0)

    return {'x': x, 'y_fit': y_fit, 'y_obs': y, 'ci_lower': ci_lower, 'ci_upper': ci_upper}

fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

for idx, (name, series) in enumerate(monthly_series.items()):
    ax = axes[idx]
    result = trend_with_bootstrap(series, n_bootstrap=500, ci=0.95)

    ax.plot(series.index[result['x']], result['y_obs'], 'o', color='lightblue', markersize=2, alpha=0.4)
    ax.plot(series.index[result['x']], result['y_fit'], 'r-', linewidth=2, label='Spline trend')
    ax.fill_between(series.index[result['x']],
                     result['ci_lower'], result['ci_upper'],
                     alpha=0.2, color='red', label=f'IC 95% bootstrap')

    # Check if CI excludes flat trend (zero slope)
    first_val = result['y_fit'][0]
    last_val = result['y_fit'][-1]
    delta = last_val - first_val
    ci_slope_lower = result['ci_lower'][-1] - result['ci_upper'][0]
    ci_slope_upper = result['ci_upper'][-1] - result['ci_lower'][0]

    ax.set_title(f'{name} — Spline trend: delta={delta:.1f} mm, IC=[{ci_slope_lower:.1f}, {ci_slope_upper:.1f}]', fontsize=12)
    ax.set_ylabel('Pluja (mm)')
    ax.legend(fontsize=8)
    ax.grid(alpha=0.3)

plt.xlabel('Data')
plt.tight_layout()
plt.show()
```

```python
# === 7.3 SARIMA with trend ===
from statsmodels.tsa.statespace.sarimax import SARIMAX

def fit_sarima_trend(series, order=(1,0,1), seasonal_order=(1,0,1,12)):
    """Fit SARIMA and extract trend from smoothed series."""
    try:
        model = SARIMAX(series.fillna(method='ffill'),
                       order=order, seasonal_order=seasonal_order,
                       enforce_stationarity=False, enforce_invertibility=False)
        res = model.fit(disp=False, maxiter=200)
        trend = res.smoothed_trend
        # MK on trend
        mk = mann_kendall(trend.values)
        ss = sens_slope(trend.values)
        return {'trend': trend, 'mk': mk, 'ss': ss, 'aic': res.aic, 'bic': res.bic}
    except Exception as e:
        return None

sarima_results = {}
for name, series in monthly_series.items():
    sr = fit_sarima_trend(series)
    sarima_results[name] = sr
    if sr:
        print(f"{name}: AIC={sr['aic']:.1f}, MK tau={sr['mk']['tau']:.4f}, p={sr['mk']['p_value']:.4f} ({sr['mk']['trend']})")
    else:
        print(f"{name}: SARIMA no va convergir")
```

```python
# SARIMA Trend plot
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)
for idx, (name, series) in enumerate(monthly_series.items()):
    ax = axes[idx]
    ax.plot(series.index, series.values, 'o-', color='lightblue', markersize=2, alpha=0.4, label='Original')
    if sarima_results[name]:
        trend = sarima_results[name]['trend']
        ax.plot(trend.index, trend.values, 'r-', linewidth=2, label='SARIMA trend')
        mk = sarima_results[name]['mk']
        ax.set_title(f'{name} — SARIMA: tau={mk["tau"]:.3f}, p={mk["p_value"]:.4f} ({mk["trend"]})', fontsize=12)
    else:
        ax.set_title(f'{name} — SARIMA: no va convergir', fontsize=12)
    ax.set_ylabel('Pluja (mm)')
    ax.legend(fontsize=8)
    ax.grid(alpha=0.3)
plt.xlabel('Data')
plt.tight_layout()
plt.show()
```
